```python
from lazypredict.Supervised import LazyRegressor
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.utils import shuffle
import numpy as np
import pandas as pd
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
diabetes = datasets.load_diabetes()
```

```python
type(diabetes)
```

```python
diabetes.DESCR
```

```python
diabetes.data
```

```python
diabetes.target
```

```python
X, y = shuffle(diabetes.data, diabetes.target, random_state=13)
```

```python
X = X.astype(np.float32)
```

```python
offset = int(X.shape[0] * 0.9)

X_train , y_train = X[:offset], y[:offset]
X_test, y_test = X[offset:], y[offset:]
```

```python
X_train
```

```python
y_train
```

```python
X[0]
```

```python
y[0]
```

```python
reg = LazyRegressor(verbose =0, ignore_warnings=False, custom_metric=None)
```

```python
models, predictions = reg.fit(X_train, X_test, y_train, y_test)
```

```python
print(models)
```

### LazyPredict ile Hızlı Model Karşılaştırması (Benchmarking)

* **Amaç:** Tek bir algoritmaya bağlı kalmadan, veri seti için en uygun regresyon modelini hızlıca tespit etmek.
* **Veri Hazırlığı:** `load_diabetes` veri seti `shuffle` ile karıştırılmış ve **%90 Eğitim / %10 Test** olarak bölünmüştür.
* **`LazyRegressor` İşlevi:** Linear, Ridge, Lasso, Random Forest, LightGBM ve SVR gibi 40'tan fazla modeli varsayılan ayarlarıyla otomatik olarak eğitir.
* **Çıktı (`models`):** Modellerin **Adjusted R-Squared**, **R-Squared**, **RMSE** ve **çalışma sürelerini** karşılaştırmalı bir tablo halinde sunar. Bu tablo, hiperparametre optimizasyonu (tuning) yapılacak en iyi 2-3 adayı seçmek için başlangıç noktasıdır.
